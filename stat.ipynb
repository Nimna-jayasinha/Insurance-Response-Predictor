```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

numeric_features = [
    "Age",
    "Annual_Premium",
    "Vintage"
]

categorical_features = [
    "Gender",
    "Driving_License",
    "Region_Code",
    "Previously_Insured",
    "Vehicle_Age",
    "Vehicle_Damage",
    "Policy_Sales_Channel"
]

preprocessor_final = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numeric_features
        ),
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first"
            ),
            categorical_features
        )
    ]
)

model_final = Pipeline(
    steps=[
        ("preprocessor", preprocessor_final),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                random_state=42
            )
        )
    ]
)
```

```python
model_final.fit(X_train, y_train)
```

```python
val_prob = model_final.predict_proba(X_val)[:, 1]

print("Min probability:", val_prob.min())
print("Max probability:", val_prob.max())
print("Mean probability:", val_prob.mean())
```

```python
from sklearn.metrics import precision_recall_curve
import numpy as np

precision, recall, thresholds = precision_recall_curve(
    y_val,
    val_prob
)

f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-10)
)

best_index = np.argmax(f1_scores)

best_threshold = thresholds[best_index]
best_precision = precision[best_index]
best_recall = recall[best_index]
best_f1 = f1_scores[best_index]

print("Best validation threshold:", best_threshold)
print("Precision:", best_precision)
print("Recall:", best_recall)
print("F1:", best_f1)
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score
import numpy as np
import pandas as pd
import warnings

warnings.filterwarnings(
    "ignore",
    message="Found unknown categories.*",
    category=UserWarning
)

C_values = [10, 1, 0.1, 0.01]

lasso_results = []

for C in C_values:
    model = Pipeline([
        ("preprocessor", preprocessor_final),
        (
            "classifier",
            LogisticRegression(
                solver="liblinear",
                l1_ratio=1.0,
                C=C,
                max_iter=1000,
                random_state=42
            )
        )
    ])

    model.fit(X_train, y_train)

    val_prob = model.predict_proba(X_val)[:, 1]

    roc_auc = roc_auc_score(y_val, val_prob)
    pr_auc = average_precision_score(y_val, val_prob)

    coefficients = model.named_steps["classifier"].coef_[0]
    total_coefficients = len(coefficients)
    zero_coefficients = np.sum(np.isclose(coefficients, 0))
    nonzero_coefficients = total_coefficients - zero_coefficients

    lasso_results.append({
        "C": C,
        "ROC_AUC": roc_auc,
        "PR_AUC": pr_auc,
        "Total_Coefficients": total_coefficients,
        "Nonzero_Coefficients": nonzero_coefficients,
        "Zero_Coefficients": zero_coefficients
    })

lasso_results_df = pd.DataFrame(lasso_results)

print(lasso_results_df)
```

```python
lasso_final = Pipeline([
    ("preprocessor", preprocessor_final),
    (
        "classifier",
        LogisticRegression(
            solver="liblinear",
            C=1.0,
            max_iter=1000,
            random_state=42,
            l1_ratio=1.0
        )
    )
])

lasso_final.fit(X_train, y_train)

# Get transformed feature names
feature_names = lasso_final.named_steps[
    "preprocessor"
].get_feature_names_out()

# Get coefficients
lasso_coef = lasso_final.named_steps[
    "classifier"
].coef_[0]

lasso_features = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": lasso_coef
})

# Separate retained and removed coefficients
retained_features = lasso_features[
    ~np.isclose(lasso_features["Coefficient"], 0)
].copy()

removed_features = lasso_features[
    np.isclose(lasso_features["Coefficient"], 0)
].copy()

print("Total coefficients:", len(lasso_features))
print("Retained:", len(retained_features))
print("Removed:", len(removed_features))
```

```python
import streamlit as st
import pandas as pd
import joblib

# --------------------------------------------------
# Load final model
# --------------------------------------------------

model = joblib.load(
    "insurance_response_model_no_gender.pkl"
)

THRESHOLD = 0.22485247334177014


# --------------------------------------------------
# Page configuration
# --------------------------------------------------

st.set_page_config(
    page_title="Insurance Customer Response Predictor",
    page_icon="📊",
    layout="centered"
)

st.title("Insurance Customer Response Predictor")

st.write(
    """
    This decision-support tool estimates a customer's likelihood
    of responding positively to a vehicle-insurance cross-sell offer.
    """
)

st.info(
    "The model does not use Gender as a predictor because removing "
    "Gender produced negligible change in predictive performance."
)


# --------------------------------------------------
# Customer information
# --------------------------------------------------

st.subheader("Customer Information")

col1, col2 = st.columns(2)

with col1:

    age = st.number_input(
        "Age",
        min_value=20,
        max_value=85,
        value=35
    )

    driving_license = st.selectbox(
        "Driving License",
        ["Yes", "No"]
    )

    previously_insured = st.selectbox(
        "Previously Insured",
        ["No", "Yes"]
    )

    vehicle_age = st.selectbox(
        "Vehicle Age",
        ["< 1 Year", "1-2 Year", "> 2 Years"]
    )

    vehicle_damage = st.selectbox(
        "Vehicle Damage",
        ["No", "Yes"]
    )


with col2:

    annual_premium = st.number_input(
        "Annual Premium",
        min_value=0.0,
        value=30000.0,
        step=1000.0
    )

    vintage = st.number_input(
        "Vintage (days)",
        min_value=10,
        max_value=299,
        value=150
    )

    region_code = st.number_input(
        "Region Code",
        min_value=0.0,
        value=28.0,
        step=1.0
    )

    policy_sales_channel = st.number_input(
        "Policy Sales Channel",
        min_value=0.0,
        value=152.0,
        step=1.0
    )


# --------------------------------------------------
# Prediction
# --------------------------------------------------

if st.button(
    "Predict Customer Response",
    type="primary",
    use_container_width=True
):

    customer = pd.DataFrame({
        "Age": [int(age)],
        "Driving_License": [
            1 if driving_license == "Yes" else 0
        ],
        "Region_Code": [float(region_code)],
        "Previously_Insured": [
            1 if previously_insured == "Yes" else 0
        ],
        "Vehicle_Age": [vehicle_age],
        "Vehicle_Damage": [vehicle_damage],
        "Annual_Premium": [float(annual_premium)],
        "Policy_Sales_Channel": [
            float(policy_sales_channel)
        ],
        "Vintage": [int(vintage)]
    })

    probability = model.predict_proba(
        customer
    )[0, 1]

    prediction = int(
        probability >= THRESHOLD
    )

    st.divider()

    st.subheader("Prediction Result")

    st.metric(
        "Model-Estimated Response Probability",
        f"{probability:.1%}"
    )

    st.caption(
        f"Decision threshold: {THRESHOLD:.1%}"
    )

    if prediction == 1:

        st.success(
            "Potential Responder — Prioritize for "
            "Cross-Sell Campaign"
        )

        st.write(
            "The customer's estimated response probability "
            "is above the selected decision threshold."
        )

    else:

        st.warning(
            "Lower-Priority Customer"
        )

        st.write(
            "The customer's estimated response probability "
            "is below the selected decision threshold."
        )


# --------------------------------------------------
# Model information
# --------------------------------------------------

with st.expander("About the Model"):

    st.write(
        """
        **Model:** HistGradientBoostingClassifier

        **Decision threshold:** 22.49%

        **Holdout ROC-AUC:** 0.8580

        **Holdout PR-AUC:** 0.3720

        **Responder Recall:** 76.44%

        **Responder Precision:** 32.79%

        **Responder F1 Score:** 45.89%
        """
    )

    st.write(
        """
        The decision threshold was selected using the
        validation dataset by maximizing the F1 score.

        Probability calibration was also evaluated using
        a calibration curve and Brier score. Predicted
        probabilities were reasonably aligned with observed
        response rates.
        """
    )


st.divider()

st.caption(
    "Decision-support prototype for academic purposes. "
    "Predictions should support, not replace, business judgment."
)
```
